```python
import pandas as pd

# Lê o arquivo original exportado do MT5
df = pd.read_csv("../raw/PETR4_M15.csv", sep='\t')

# Converte DATE + TIME para coluna única Date
df["Date"] = pd.to_datetime(
    df["<DATE>"] + " " + df["<TIME>"],
    format="%Y.%m.%d %H:%M:%S",
    errors="coerce"
)

# Mapeamento para nomes limpos
rename_map = {
    "<OPEN>": "open",
    "<HIGH>": "high",
    "<LOW>": "low",
    "<CLOSE>": "close",
    "<TICKVOL>": "tickvol",
    "<VOL>": "volume",
    "<SPREAD>": "spread"
}

# Renomeia apenas colunas que existem (evita erro se faltar alguma)
df = df.rename(columns={col: rename_map[col] for col in df.columns if col in rename_map})

# Seleciona colunas relevantes (somente as que existem)
cols_to_keep = ["Date"]
for c in ["open", "high", "low", "close", "tickvol", "volume", "spread"]:
    if c in df.columns:
        cols_to_keep.append(c)

df_out = df[cols_to_keep].copy()

# Ordena por data
df_out = df_out.sort_values("Date").reset_index(drop=True)

# Salva no processed
df_out.to_csv("../processed/PETR4_M15.csv", index=False)

print("Arquivo limpo salvo em ../processed/PETR4_M15.csv")
print("Colunas finais:", df_out.columns.tolist())
print("Total de linhas:", len(df_out))
```
