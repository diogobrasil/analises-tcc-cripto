```python
import os
import sys
sys.path.append(os.path.abspath("../../.."))

import pandas as pd
import matplotlib.pyplot as plt

from classes.neural_networks.training.train_arima import (
    train_and_evaluate_arima,
    load_series,
    split_series
)
```

```python
# --- Parâmetros ---
csv_path   = "../../datasets/b3_dados/processed/acoes_concat.csv"
target     = "ITUB4"
order      = (2, 1, 1)
model_dir  = "../../saved_models/arima"
version    = "1.0"
train_end  = "2017-12-31"
val_end    = "2018-12-31"
rolling    = True
step_size  = 1
```

```python
# --- Carrega dados e divide em train/val/test ---
ts    = load_series(csv_path, target)
train, val, test = split_series(ts, train_end, val_end)

# --- Treina e avalia ---
result = train_and_evaluate_arima(
    csv_path=csv_path,
    target=target,
    order=order,
    model_dir=model_dir,
    version=version,
    train_end=train_end,
    val_end=val_end,
    rolling=rolling,
    step_size=step_size
)

metrics     = result["metrics"]
predictions = result["predictions"]
train_pred  = predictions["train"]
val_pred    = predictions["val"]
test_pred   = predictions["test"]

# --- Exibe métricas ---
print("=== TRAIN METRICS ===")
for name, val in metrics["train"].items():
    print(f"{name}: {val:.4f}")

print("\n=== VAL METRICS ===")
for name, val in metrics["val"].items():
    print(f"{name}: {val:.4f}")

print("\n=== TEST METRICS ===")
for name, val in metrics["test"].items():
    print(f"{name}: {val:.4f}")
```
